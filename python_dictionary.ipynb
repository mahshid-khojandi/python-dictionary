dictionaries are indexed by keys, which can be any immutable type; strings and numbers can always be keys.

You can’t use lists as keys

A pair of braces creates an empty dictionary: {}.

dictionary: { key: value }



```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
print("Dave's age : ", ages["Dave"])
print("Mary's age : ", ages["Mary"])
ages["Mary"] = 15
print("Mary's age : ", ages["Mary"])
ages['kevin'] = 28
print(ages)
```

It is also possible to delete a key:value pair with del

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
del ages["Mary"]
print(ages)
```

To check whether a single key is in the dictionary, use the in keyword.

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
'Mary' in ages
'a' not in ages
```

When looping through dictionaries, the key and corresponding value can be retrieved at the same time using the items() method.

```python
numbers = {'a': 1 , 'b': 2}
for k, v in numbers.items():
    print(k, v)
```

The clear method is used to empty the contents of a dictionary

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
ages.clear()
print(ages)
```

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
ages.get("Dave")
```

The keys method is used to retrieve all the keys of a dictionary

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
ages.keys()
```

The values ​​method is used to retrieve all the values ​​of a dictionary

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
ages.values()
```

The pop method removes a key from the dictionary and returns the removed value

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
x = ages.pop('Dave')
print(x)
```

The popitem method removes the last item from the dictionary and returns the removed value

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
x = ages.popitem()
print (x)
print (ages)
```

The update method is used to combine one dictionary with another dictionary

If the key is not already in the dictionary, a new key will be created

```python
ages = {"Dave": 24, "Mary": 42, "John": 58}
tel = {'jack': 4098, 'sape': 4139}
ages.update(tel)
print(ages)
```
